```python
from huggingface_hub import hf_hub_download
from pathlib import Path
import shutil

REPO_ID = "CS2CD/CS2CD.Counter-Strike_2_Cheat_Detection"
DATA_FOLDER = Path("datasets/test")

DATA_FOLDER.mkdir(parents=True, exist_ok=True)

for i in range(56, 60):
    for extension in ["parquet", "json"]:
        filename = f"{i}.{extension}"

        try:
            downloaded_file = hf_hub_download(
                repo_id=REPO_ID,
                filename=f"with_cheater_present/{filename}",
                repo_type="dataset"
            )

            destination = DATA_FOLDER / filename
            shutil.copy2(downloaded_file, destination)

            print(f"Downloaded: {filename}")

        except Exception as e:
            print(f"Failed to download {filename}: {e}")

print("Download process completed!")
```
