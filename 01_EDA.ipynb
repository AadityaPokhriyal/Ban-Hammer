```python
cheater_ids = {
    player["steamid"]
    for player in events["cheaters"]
}

df["is_cheater"] = (
    df["steamid"].isin(cheater_ids)
).astype("int8")

print(df[["steamid", "is_cheater"]].drop_duplicates())
```

```python
print("Cheaters:", cheater_ids)

print("\nPlayers in Parquet:")
print(sorted(df["steamid"].dropna().unique()))

print("\nRows per player:")
print(df["steamid"].value_counts())

print("\nWarmup distribution:")
print(df["is_warmup_period"].value_counts(dropna=False))
```
